```python
# Imports and load raw data

import pandas as pd
import numpy as np

df = pd.read_csv("../data/Raw datasets/Ev_registration_car.csv")

print("Chargers shape:", df.shape)
df.head()
```

```python
# Create a working copy

ev_car_df = df.copy()

print("Original shape:", df.shape)
print("Working copy shape:", ev_car_df.shape)
```

```python
# Rename State column

ev_car_df = ev_car_df.rename(columns={
    ev_car_df.columns[0]: "State"
})

print(ev_car_df.columns.tolist())
```

```python
# Remove Total row

ev_car_df = ev_car_df[ev_car_df["State"] != "Total"].copy()

print("Shape after removing Total row:", ev_car_df.shape)
```

```python
# Check Duplicate States

print("Duplicate state names:", ev_car_df["State"].duplicated().sum())
```

```python
# Check Missing Values

print(ev_car_df.isnull().sum())
```

```python
# Check data types

print(ev_car_df.dtypes)
```

```python
# Check Negative Values

numeric_cols = ev_car_df.columns[1:]
print((ev_car_df[numeric_cols] < 0).sum())
```

```python
# Validate Total Column

year_cols = [str(year) for year in range(2017, 2027)]

calculated_total = ev_car_df[year_cols].sum(axis=1)

print(
    "Rows where Total does not match calculated total:",
    (ev_car_df["Total"] != calculated_total).sum()
)
```

```python
# Check Duplicate Rows

print("Duplicate rows:", ev_car_df.duplicated().sum())
```

```python
print(
    "States with leading/trailing whitespace:",
    (~ev_car_df["State"].eq(ev_car_df["State"].str.strip())).sum()
)
```

```python
print(
    "Blank State values:",
    ev_car_df["State"].str.strip().eq("").sum()
)
```

```python
print((ev_car_df[year_cols] == 0).sum())
```

```python
zero_check = ev_car_df[year_cols].eq(0)

display(
    ev_car_df.loc[
        zero_check.any(axis=1),
        ["State"] + year_cols
    ].reset_index(drop=True)
)
```

```python
# SUMMARY STATISTICS

ev_car_df.describe()
```

```python
# Save cleaned car EV dataset

clean_path = "../data/Cleaned datasets/Ev_registration_car_cleaned.csv"

ev_car_df.to_csv(clean_path, index=False)

print("Clean dataset saved successfully.")
print("Shape:", ev_car_df.shape)
```
