```python
# Imports and load raw data

import pandas as pd
import numpy as np

df = pd.read_csv("../data/Raw datasets/Ev_registration_bus.csv")

print("Chargers shape:", df.shape)
df.head()
```

```python
# Create a working copy

ev_bus_df = df.copy()

print("Original shape:", df.shape)
print("Working copy shape:", ev_bus_df.shape)
```

```python
# Rename first column

ev_bus_df = ev_bus_df.rename(columns={ev_bus_df.columns[0]: "State"})

print(ev_bus_df.columns.tolist())
```

```python
# Remove Total row

ev_bus_df = ev_bus_df[ev_bus_df["State"] != "Total"].copy()

print("Shape after removing Total row:", ev_bus_df.shape)
```

```python
# Check duplicate states

duplicate_states = ev_bus_df["State"].duplicated().sum()

print("Duplicate state names:", duplicate_states)
```

```python
# Check missing values

print(ev_bus_df.isnull().sum())
```

```python
# Check data types

print(ev_bus_df.dtypes)
```

```python
# Check negative values

year_cols = [str(year) for year in range(2017, 2027)]

print((ev_bus_df[year_cols] < 0).sum())
```

```python
# Validate Total Column

calculated_total = ev_bus_df[year_cols].sum(axis=1)

print(
    "Rows where Total does not match calculated total:",
    (ev_bus_df["Total"] != calculated_total).sum()
)
```

```python
# Check duplicate rows
print("Duplicate rows:", ev_bus_df.duplicated().sum())
```

```python
# Check leading/trailing whitespace

whitespace_rows = ev_bus_df[
    ev_bus_df["State"] != ev_bus_df["State"].str.strip()
]

print("State names with extra whitespace:", len(whitespace_rows))
```

```python
# Check blank state names

print("Blank state names:", ev_bus_df["State"].str.strip().eq("").sum())
```

```python
# Count zero registrations by year

print((ev_bus_df[year_cols] == 0).sum())
```

```python
# View rows containing at least one zero

zero_rows = ev_bus_df[
    (ev_bus_df[year_cols] == 0).any(axis=1)
]

display(
    zero_rows[["State"] + year_cols].reset_index(drop=True)
)
```

```python
# Summary statistics

display(
    ev_bus_df[year_cols + ["Total"]].describe()
)
```

```python
# Save cleaned dataset

clean_path = "../data/Cleaned datasets/Ev_registration_bus_cleaned.csv"

ev_bus_df.to_csv(clean_path, index=False)

print("Clean dataset saved successfully.")
print("Shape:", ev_bus_df.shape)
```
